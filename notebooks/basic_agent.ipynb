```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def addNumbers(a:int, b:int):
    """
        it will return the sum of two numbers
        Args: 
            a: Number One
            b: Number Two
    """
    return a+b

@tool
def productNumbers(c:int, d:int):
    """
        it will return product of two numbers
        Args:
            c: Number One
            d: Number Two
    """
    return c*d
```

```python
addNumbers.invoke({"a":12, "b":14})
productNumbers.invoke({"c":5, "d":14})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
agent=create_agent(
    model=llm,
    tools=[addNumbers,productNumbers],
    system_prompt="You are a math teacher and always use tools for calculations"
)
```

```python
response= agent.invoke({"messages":[{"role":"user", "content": "What is 2+3?"}]})
```
